```python
import time
def time_model(pipe, n_repeats=5, n_pred=1000):
    """Median fit time and per-sample predict time. Set every model's n_jobs=1 first."""
    fit_t, pred_t = [], []
    for _ in range(n_repeats):
        t0 = time.perf_counter(); pipe.fit(X_train, y_train); fit_t.append(time.perf_counter() - t0)
        t0 = time.perf_counter(); pipe.predict(X_train.iloc[:n_pred]); pred_t.append(time.perf_counter() - t0)
    ms_per_sample = 1000 * np.median(pred_t) / n_pred
    return {"fit_s": np.median(fit_t), "predict_ms_per_sample": ms_per_sample,
            "meets_C2": ms_per_sample <= LATENCY_LIMIT_MS, "runtime": RUNTIME}

# TODO: paste each owner's best pipeline (best_params) here, time it, add model size (pickle bytes)
# and a training-size scalability curve.
# FINAL STEP (once, after the group agrees on the recommendation):
# final.fit(X_train, y_train); y_hat = final.predict(X_test); report MAE and pct_within(y_test, y_hat)
```
